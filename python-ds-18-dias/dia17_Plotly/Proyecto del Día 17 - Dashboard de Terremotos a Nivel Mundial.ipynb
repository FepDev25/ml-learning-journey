## Cargar y preparar datos

```python
import pandas as pd
import plotly.graph_objects as go
import plotly.express as px
import dash
from dash import dcc, html
import pandas as pd
import plotly.express as px
```

```python
df = pd.read_csv("./earthquake+data.csv")
df.head(5)
```

```python
df.info()
```

```python
df["Date & Time"] = pd.to_datetime(df["Date & Time"])
```

```python
df.info()
```

## Creación de gráficas

### 1. Mapa Coroplético: Un mapa coroplético que muestre la densidad de terremotos por región.

```python
fig1 = px.choropleth(
    df,
    locations="Country",
    locationmode="country names",
    color="Magnitude",
    hover_name="Country",
    projection="natural earth",
)
fig1.update_layout(
    title="Earthquake Magnitude by Country"
)
```

### 2. Gráfico de Barras: Un gráfico de barras que muestre el número de terremotos por magnitud.

```python
magnitudes_count = df['Magnitude'].value_counts().sort_index()

# Numero de terremotos por magnitud
fig2 = go.Figure()
fig2.add_trace( go.Bar(
    x=df["Magnitude"].unique(),
    y=magnitudes_count,
    marker={
        'color': magnitudes_count.values,
        'colorscale': 'Viridis',
        'colorbar' : {
            'title': 'Número de Terremotos',
        },
        'line': {
            'width': 1.5,
            'color': 'black'
        },
    },
    text= magnitudes_count.values,
    textposition= 'outside'
) 
)

fig2.update_layout(
    title="Número de Terremotos por Magnitud",
    xaxis={
        'title': "Magnitud",
        'tickmode': 'linear',
        'dtick': 1
    },
    yaxis={
        'title': "Número de Terremotos"
    }
)
```

### 3. Gráfico de Líneas: Un gráfico de líneas que muestre la tendencia de la magnitud de los terremotos a lo largo del tiempo.

```python
df.sort_values(by="Date & Time", inplace=True)

fig3 = go.Figure()
fig3.add_trace(
    go.Scatter(
        x=df["Date & Time"],
        y=df["Magnitude"],
        mode="lines"
    )
)
fig3.update_layout(
    title="Terremotos a lo largo del tiempo",
    xaxis={
        'title': "Fecha",
        'tickformat': '%Y-%m-%d %H:%M:%S',
    },
    yaxis={
        'title': "Magnitud"
    }
)
```

### 4. Gráfico de Dispersión: Un gráfico de dispersión que muestre la relación entre la magnitud de los terremotos y su profundidad.

```python
fig4 = go.Figure( data=go.Scatter(
    x=df["Depth"],
    y=df["Magnitude"],
    mode='markers',
    marker={
        'color': df["Magnitude"],
        'colorscale': 'Viridis',
        'size': 5,
        'colorbar': {
            'title': 'Magnitud',
        },
    }
) )
fig4.update_layout(
    title='Magnitud vs Profundidad',
    xaxis_title='Profundidad',
    yaxis_title='Magnitud',
)
```

### 5. Gráfico 3D: Un gráfico 3D que visualice la relación entre magnitud, profundidad y tiempo.

```python
fig5 = go.Figure(
    data=[go.Scatter3d(
        x=df["Magnitude"],
        y=df["Depth"],
        z=df["Date & Time"],
        mode='markers',
        marker={
            'size': 10,
            'color': df["Magnitude"],
            'colorscale': 'Viridis',
            'opacity': 0.8,
        }
    )]
)
fig5.update_layout(
    title='Terremotos en 3D',
    scene=dict(
        xaxis_title='Magnitud',
        yaxis_title='Profundidad',
        zaxis_title='Fecha')
)
```

### Diseño de aplicación

```python
app = dash.Dash(__name__)

app.layout = html.Div(
    children=[
        html.H1(children='Visualizaciiones de Terremotos'),

        dcc.Graph(
            id='fig1',
            figure=fig1
        ),

        dcc.Graph(
            id='fig2',
            figure=fig2
        ),

        dcc.Graph(
            id='fig3',
            figure=fig3
        ),

        dcc.Graph(
            id='fig4',
            figure=fig4
        ),

        dcc.Graph(
            id='fig5',
            figure=fig5
        )
    ]
)
```

```python
if __name__ == "__main__":
    app.run(debug=True)
```
